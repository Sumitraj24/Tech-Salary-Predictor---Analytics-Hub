# Remote Job Scraping & Data Collection
This notebook fetches real-time tech job listings and salary data from public developer job APIs:
- **RemoteOK API** ()
- **Remotive API** ()

Outputs are structured and exported to CSV files for data mining and market analysis.

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

url = "https://remoteok.com/api"

data = requests.get(url).json()

jobs = []
count = 0

for job in data[1:]:  # skip metadata

    jobs.append({
        "title": job.get("position"),
        "company": job.get("company"),
        "tags": job.get("tags"),
        "salary": job.get("salary")
    })

    count += 1

    # ✅ print progress every 50 rows
    if count % 50 == 0:
        print(f"{count} rows scraped... continuing")

# ✅ save to CSV
df = pd.DataFrame(jobs)
df.to_csv("remoteok_jobs.csv", index=False)

print(f"\n✅ Done! Total {count} jobs saved to remoteok_jobs.csv")
```

```python
import requests
import pandas as pd

url = "https://remotive.com/api/remote-jobs"

response = requests.get(url)
data = response.json()

jobs = []

for job in data['jobs']:
    jobs.append({
        "title": job["title"],
        "company": job["company_name"],
        "location": job["candidate_required_location"],
        "salary": job["salary"]
    })

df = pd.DataFrame(jobs)

df.to_csv("jobs_data.csv", index=False)

print(df.head())
print("Total jobs:", len(df))
```
